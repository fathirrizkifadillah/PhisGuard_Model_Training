## 1.10 Lexical Diversity & Spurious Corpus Bias (Enron vs Nazario)

```python
def get_top_tokens(texts, n=12):
    stopwords = {'the', 'to', 'and', 'of', 'a', 'in', 'for', 'is', 'on', 'that', 'this', 'you', 'i', 'it', 'with', 'as', 'be', 'are', 'at', 'from', 'by', 'have', 'or', 'an', 'was', 'we', 'will', 'my', 'if', 'not', 'but', 'all', 'your', 'has', 'would', 'can', 'so', 'he', 'one', 'they', 'about', 'which', 'our', 'out', 'up', 'more', 'no', 'when', 'what', 'who', 'any', 'their', 'there', 'some', 'me', 'were', 'than', 'had', 'been', 'do', 'them', 're', 's', 't', 'd', 'm', 'o', 'subject', 'http', 'com', 'www', 'font', 'td', 'tr', 'br'}
    words = []
    for t in texts:
        tokens = re.findall(r'\b[a-z]{3,}\b', str(t).lower())
        words.extend([w for w in tokens if w not in stopwords])
    return pd.DataFrame(Counter(words).most_common(n), columns=['Token', 'Count'])

safe_words = get_top_tokens(df_analysis[df_analysis['Email Type']=='Safe Email']['Email Text'], n=12)
phish_words = get_top_tokens(df_analysis[df_analysis['Email Type']=='Phishing Email']['Email Text'], n=12)

print("Top 12 Token Safe Email:")
display(safe_words)
print("Top 12 Token Phishing Email:")
display(phish_words)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
axes[0].barh(safe_words['Token'][::-1], safe_words['Count'][::-1], color='#3498db', edgecolor='black', alpha=0.85)
axes[0].set_title('Top Distinctive Tokens in Safe Email\n(Notice Enron enterprise terms: enron, deal, gas, ect)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Token Frequency')

axes[1].barh(phish_words['Token'][::-1], phish_words['Count'][::-1], color='#e74c3c', edgecolor='black', alpha=0.85)
axes[1].set_title('Top Distinctive Tokens in Phishing Email\n(Notice action/social-engineering terms: click, email, money, free)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Token Frequency')

plt.suptitle('Lexical Domain Differences (Corpus Bias vs Phishing Cues)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plot_file5 = PLOT_DIR / '05_email_top_words_comparison.png'
plt.savefig(plot_file5, dpi=200, bbox_inches='tight')
plt.show()
```

## 1.11 Statistical Hypothesis Testing (Mann-Whitney U Test)

```python
u_char, p_char = stats.mannwhitneyu(
    df_analysis[df_analysis['Email Type']=='Phishing Email']['char_length'],
    df_analysis[df_analysis['Email Type']=='Safe Email']['char_length'],
    alternative='two-sided'
)

u_word, p_word = stats.mannwhitneyu(
    df_analysis[df_analysis['Email Type']=='Phishing Email']['word_count'],
    df_analysis[df_analysis['Email Type']=='Safe Email']['word_count'],
    alternative='two-sided'
)

print(f"Mann-Whitney U Test - Character Length: U-stat = {u_char:.2e}, p-value = {p_char:.2e}")
print(f"Mann-Whitney U Test - Word Count:       U-stat = {u_word:.2e}, p-value = {p_word:.2e}")
```

# CHAPTER 2 : Data Cleaning & Text Preprocessing

Pada bab ini, kita mengeksekusi pipeline *data cleanroom* dan rekayasa teks NLP untuk menghasilkan korpus bersih yang siap diproses oleh TF-IDF Vectorizer:
1. **Eliminasi Missing Value & Whitespace:** Menghapus 16 baris `NaN` dan baris yang hanya berisi whitespace/newline.
2. **Eliminasi Konflik Label Template:** Menghapus 533 baris teks literal `"empty"` yang menimbulkan noise label fatal.
3. **Filtering Outlier Ekstrim:** Menghapus baris 12500 (dump notebook 17 MB) dan membatasi teks maksimal $\le 50.000$ karakter.
4. **Deduplikasi Korpus:** Menghapus teks duplikat untuk mencegah kebocoran informasi (*data leakage*) antara train dan test set.
5. **Pembersihan Kolom Redundan:** Mengeliminasi kolom index CSV `Unnamed: 0`.
6. **Standardisasi Label Target:** Melakukan encoding label biner menjadi `is_phishing` ($1 = \text{Phishing Email}$, $0 = \text{Safe Email}$).
7. **Rekayasa Pembersihan Teks NLP:** Membangun fungsi modular pembersihan teks:
   * Stripping tag HTML (`<...>` $\rightarrow$ spasi).
   * Normalisasi tautan hyperlink menjadi token khusus `url_token`.
   * Normalisasi alamat email menjadi token khusus `email_token`.
   * Lowercasing dan normalisasi spasi.
8. **Validasi Integritas Sebelum vs Sesudah:** Memverifikasi kualitas korpus akhir.
9. **Penyimpanan Data Bersih:** Mengekspor hasil pembersihan ke `data/email/processed/email_cleaned.csv`.
10. **Pembagian Data Bebas Leakage (Train-Test Split):** Melakukan stratified train-test split ($80:20$) dengan `random_state=42` sesuai panduan scikit-learn.


## 2.1 Eksekusi Pipeline Data Cleanroom Korpus Email

```python
email_audit_tracker = []
email_audit_tracker.append({'Tahap': '1. Dataset Mentah (Raw Ingestion)', 'Jumlah Baris': len(df_raw), 'Baris Dihapus': 0})

# 1. Hapus kolom index redundan 'Unnamed: 0'
df_cleaned_email = df_raw.copy()
if 'Unnamed: 0' in df_cleaned_email.columns:
    df_cleaned_email = df_cleaned_email.drop(columns=['Unnamed: 0'])

# 2. Hapus baris dengan Email Text NaN
n_before = len(df_cleaned_email)
df_cleaned_email = df_cleaned_email.dropna(subset=['Email Text']).copy()
dropped_null = n_before - len(df_cleaned_email)
email_audit_tracker.append({'Tahap': '2. Eliminasi Email Text NaN', 'Jumlah Baris': len(df_cleaned_email), 'Baris Dihapus': dropped_null})

# 3. Hapus baris teks kosong atau whitespace-only
n_before = len(df_cleaned_email)
df_cleaned_email = df_cleaned_email[df_cleaned_email['Email Text'].str.strip() != ''].copy()
dropped_whitespace = n_before - len(df_cleaned_email)
email_audit_tracker.append({'Tahap': '3. Eliminasi Teks Whitespace / Newline', 'Jumlah Baris': len(df_cleaned_email), 'Baris Dihapus': dropped_whitespace})

# 4. Hapus 533 baris teks literal 'empty' yang memicu konflik label
n_before = len(df_cleaned_email)
df_cleaned_email = df_cleaned_email[df_cleaned_email['Email Text'].str.strip().str.lower() != 'empty'].copy()
dropped_empty = n_before - len(df_cleaned_email)
email_audit_tracker.append({'Tahap': '4. Eliminasi Konflik Label (Teks Literal empty)', 'Jumlah Baris': len(df_cleaned_email), 'Baris Dihapus': dropped_empty})

# 5. Filter outlier ekstrim (> 50,000 karakter, termasuk dump notebook 17MB)
df_cleaned_email['char_length'] = df_cleaned_email['Email Text'].str.len()
n_before = len(df_cleaned_email)
df_cleaned_email = df_cleaned_email[df_cleaned_email['char_length'] <= 50000].copy()
dropped_outlier = n_before - len(df_cleaned_email)
email_audit_tracker.append({'Tahap': '5. Eliminasi Outlier Ekstrim (> 50k chars)', 'Jumlah Baris': len(df_cleaned_email), 'Baris Dihapus': dropped_outlier})

# 6. Deduplikasi teks email identik
n_before = len(df_cleaned_email)
df_cleaned_email = df_cleaned_email.drop_duplicates(subset=['Email Text']).copy()
dropped_dedup = n_before - len(df_cleaned_email)
email_audit_tracker.append({'Tahap': '6. Deduplikasi Teks Email', 'Jumlah Baris': len(df_cleaned_email), 'Baris Dihapus': dropped_dedup})

# 7. Standardisasi label target (1 = Phishing Email, 0 = Safe Email)
df_cleaned_email['is_phishing'] = (df_cleaned_email['Email Type'] == 'Phishing Email').astype(int)

# Tampilkan log ringkasan tahapan pembersihan
email_tracker_df = pd.DataFrame(email_audit_tracker)
email_tracker_df['Total Berkurang Kumulatif'] = len(df_raw) - email_tracker_df['Jumlah Baris']
print("Log Eksekusi Cleanroom Data Email:")
display(email_tracker_df)
```

## 2.2 Rekayasa Pembersihan Teks NLP (Text Preprocessing)

Membangun fungsi preprocessing teks modular untuk membersihkan tag markup dan menormalkan entitas leksikal:
* Menghapus tag HTML (`<.*?>`).
* Menormalkan hyperlink menjadi token generik `url_token`.
* Menormalkan alamat email menjadi token generik `email_token`.
* Mengubah ke huruf kecil (*lowercasing*).
* Menghapus karakter spasi dan newline berlebih.


```python
def clean_email_text(text):
    text = str(text)
    # 1. Hapus tag HTML
    text = re.sub(r'<[^>]+>', ' ', text)
    # 2. Normalisasi URL
    text = re.sub(r'http[s]?://\S+|www\.\S+', ' url_token ', text)
    # 3. Normalisasi Email
    text = re.sub(r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}', ' email_token ', text)
    # 4. Lowercase
    text = text.lower()
    # 5. Normalisasi spasi dan karakter whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Terapkan fungsi pembersihan ke seluruh korpus
print("Menjalankan pembersihan teks pada seluruh baris...")
df_cleaned_email['cleaned_text'] = df_cleaned_email['Email Text'].apply(clean_email_text)

# Hitung ulang metrik panjang karakter dan kata pasca-pembersihan
df_cleaned_email['cleaned_char_length'] = df_cleaned_email['cleaned_text'].str.len()
df_cleaned_email['cleaned_word_count'] = df_cleaned_email['cleaned_text'].apply(lambda x: len(x.split()))

# Tampilkan 3 sampel teks sebelum vs sesudah
sample_comparison = df_cleaned_email[['Email Text', 'cleaned_text', 'is_phishing']].head(3)
for idx, row in sample_comparison.iterrows():
    label_name = 'Phishing' if row['is_phishing'] == 1 else 'Safe'
    print(f"--- Sampel Baris {idx} ({label_name}) ---")
    print("Sebelum (Raw):", repr(row['Email Text'][:100]))
    print("Sesudah (Cleaned):", repr(row['cleaned_text'][:100]))
```

## 3.3 Analisis Leksikal: Top Unigram Paling Diskriminatif

Mengekstraksi kata tunggal (*unigram*) paling dominan di masing-masing kelas setelah mengeliminasi stopwords umum bahasa Inggris.


```python
# Stopwords kustom untuk analisis leksikal
custom_stopwords = {'the', 'to', 'and', 'of', 'a', 'in', 'for', 'is', 'on', 'that', 'this', 'you',
                    'i', 'it', 'with', 'as', 'be', 'are', 'at', 'from', 'by', 'have', 'or', 'an',
                    'was', 'we', 'will', 'my', 'if', 'not', 'but', 'all', 'your', 'has', 'would',
                    'can', 'so', 'he', 'one', 'they', 'about', 'which', 'our', 'out', 'up', 'more',
                    'no', 'when', 'what', 'who', 'any', 'their', 'there', 'some', 'me', 'were',
                    'than', 'had', 'been', 'do', 'them', 're', 's', 't', 'd', 'm', 'o', 'subject',
                    'http', 'com', 'www', 'font', 'td', 'tr', 'br', 'table', 'href', 'width',
                    'border', 'nbsp', 'height', 'align', 'color'}

def extract_top_unigrams(corpus, top_n=15):
    vec = CountVectorizer(stop_words=list(custom_stopwords), min_df=5, token_pattern=r'\b[a-z]{3,}\b')
    X_counts = vec.fit_transform(corpus)
    word_freq = X_counts.sum(axis=0)
    words = [(word, word_freq[0, idx]) for word, idx in vec.vocabulary_.items()]
    words = sorted(words, key=lambda x: x[1], reverse=True)
    return pd.DataFrame(words[:top_n], columns=['Token', 'Frequency'])

safe_corpus = df_clean_email[df_clean_email['is_phishing'] == 0]['cleaned_text']
phish_corpus = df_clean_email[df_clean_email['is_phishing'] == 1]['cleaned_text']

top_unigram_safe = extract_top_unigrams(safe_corpus, top_n=12)
top_unigram_phish = extract_top_unigrams(phish_corpus, top_n=12)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

axes[0].barh(top_unigram_safe['Token'][::-1], top_unigram_safe['Frequency'][::-1],
             color='#3498db', edgecolor='black', alpha=0.85)
axes[0].set_title('Top Unigrams in Safe Emails (Enterprise Communication Focus)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Token Frequency')

axes[1].barh(top_unigram_phish['Token'][::-1], top_unigram_phish['Frequency'][::-1],
             color='#e74c3c', edgecolor='black', alpha=0.85)
axes[1].set_title('Top Unigrams in Phishing Emails (Call-to-Action & Lure Focus)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Token Frequency')

plt.suptitle('Unigram Lexical Divergence: Safe vs Phishing Emails', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
eda_file2 = EDA_PLOT_DIR / 'eda_02_top_unigrams_comparison.png'
plt.savefig(eda_file2, dpi=200, bbox_inches='tight')
plt.show()
print(f"Plot disimpan di: {eda_file2}")
```

## 3.4 Analisis N-Gram: Ekstraksi Top Bigram (Frasa Kunci)

Kata tunggal sering kali ambigu. Namun, pasangan kata (*bigram*) menangkap pola rekayasa sosial secara jauh lebih spesifik, seperti `"click here"`, `"verify account"`, `"security alert"`.


```python
def extract_top_bigrams(corpus, top_n=12):
    vec = CountVectorizer(ngram_range=(2, 2), stop_words=list(custom_stopwords), min_df=5, token_pattern=r'\b[a-z]{3,}\b')
    X_counts = vec.fit_transform(corpus)
    word_freq = X_counts.sum(axis=0)
    words = [(word, word_freq[0, idx]) for word, idx in vec.vocabulary_.items()]
    words = sorted(words, key=lambda x: x[1], reverse=True)
    return pd.DataFrame(words[:top_n], columns=['Bigram Phrase', 'Frequency'])

print("Mengekstraksi bigram dari korpus...")
top_bigram_safe = extract_top_bigrams(safe_corpus, top_n=12)
top_bigram_phish = extract_top_bigrams(phish_corpus, top_n=12)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].barh(top_bigram_safe['Bigram Phrase'][::-1], top_bigram_safe['Frequency'][::-1],
             color='#3498db', edgecolor='black', alpha=0.85)
axes[0].set_title('Top Bigrams in Safe Emails\n(Corporate Dealings & Scheduling)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Phrase Frequency')

axes[1].barh(top_bigram_phish['Bigram Phrase'][::-1], top_bigram_phish['Frequency'][::-1],
             color='#e74c3c', edgecolor='black', alpha=0.85)
axes[1].set_title('Top Bigrams in Phishing Emails\n(Social Engineering & Urgency Prompts)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Phrase Frequency')

plt.suptitle('Bigram Phrase Comparison: Contextual Indicators of Phishing', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
eda_file3 = EDA_PLOT_DIR / 'eda_03_top_bigrams_comparison.png'
plt.savefig(eda_file3, dpi=200, bbox_inches='tight')
plt.show()
print(f"Plot disimpan di: {eda_file3}")
```

## 3.6 Rangkuman Wawasan EDA & Konfigurasi TF-IDF Vectorizer

Berdasarkan analisis eksplorasi data NLP di atas, berikut adalah implikasi krusial untuk tahapan Feature Engineering (Chapter 4) dan Model Training (Chapter 5):

1. **Keharusan Memasukkan N-Gram Berurutan (`ngram_range=(1,2)`):**
   * Bigram seperti `"click here"`, `"security alert"`, dan `"verify account"` memiliki korelasi jauh lebih kuat terhadap serangan phishing dibanding unigram yang berdiri sendiri. TF-IDF harus mengikutsertakan unigram dan bigram.
2. **Penerapan `sublinear_tf=True`:**
   * Teks email memiliki variasi panjang yang lebar (median 700-900 karakter, tetapi dapat mencapai puluhan ribu karakter). Mengaktifkan `sublinear_tf=True` mengganti bobot frekuensi term $TF$ menjadi $1 + \log(TF)$, sehingga kata yang diulang puluhan kali tidak mendominasi vektor secara tidak proporsional.
3. **Penyaringan Frekuensi Dokumen (`min_df=5`, `max_df=0.85`):**
   * Menetapkan `min_df=5` efektif mengeliminasi token langka atau salah ketik (*typo artifacts*).
   * Menetapkan `max_df=0.85` menyaring istilah korporat yang muncul di hampir seluruh korespondensi Enron, mencegah model mengalami overfitting.
4. **Kesesuaian Model Linear SVM (Support Vector Machine):**
   * Dengan kombinasi unigram dan bigram, ruang dimensi fitur TF-IDF akan mencapai lebih dari $25.000$ dimensi dengan matriks sangat renggang (*sparse matrix*).
   * Linear SVM (`LinearSVC` atau `SGDClassifier(loss='hinge')`) secara matematis adalah arsitektur optimal untuk klasifikasi teks berdimensi sangat tinggi karena mencari *maximum-margin separating hyperplane* secara efisien dengan kompleksitas waktu pelatihan linear terhadap jumlah dokumen.


## 3.7 Analisis Stilometrik Lanjutan (Gaya Penulisan & Pemicu Intimidasi)

Analisis stilometrik (*stylometric analysis*) berfokus pada kebiasaan sintaksis dan gaya penulisan penyerang, yang sering kali menggunakan taktik intimidasi psikologis:
1. **Uppercase Ratio (Shouting Ratio):** Proporsi huruf kapital dalam email. Penyerang phishing sering menggunakan huruf kapital berlebih untuk memicu kepanikan (contoh: "URGENT", "ACTION REQUIRED", "SUSPENDED").
2. **Exclamation Density:** Jumlah tanda seru (`!`) dalam teks.
3. **Dollar Sign Density:** Jumlah simbol mata uang (`$`) yang memicu iming-iming finansial.
4. **Digit Density:** Rasio karakter angka dalam teks.
5. **Lexical Diversity (Type-Token Ratio / TTR):** Rasio jumlah kata unik terhadap total kata, mengukur variasi kosakata.


```python
# Muat dataset mentah untuk mengekstraksi fitur stilometrik asli (sebelum lowercasing)
df_raw_styled = pd.read_csv('../data/email/raw/phishing_email.csv').dropna(subset=['Email Text'])
# Terapkan filter yang sama persis dengan cleanroom Chapter 2
df_raw_styled = df_raw_styled[df_raw_styled['Email Text'].str.strip() != ''].copy()
df_raw_styled = df_raw_styled[df_raw_styled['Email Text'].str.strip().str.lower() != 'empty'].copy()
df_raw_styled = df_raw_styled[df_raw_styled['Email Text'].str.len() <= 50000].copy()
df_raw_styled = df_raw_styled.drop_duplicates(subset=['Email Text']).copy()
df_raw_styled['is_phishing'] = (df_raw_styled['Email Type'] == 'Phishing Email').astype(int)
df_raw_styled['Class_Name'] = df_raw_styled['is_phishing'].map({0: 'Safe Email (0)', 1: 'Phishing Email (1)'})

# Ekstraksi metrik stilometrik
print("Mengekstraksi fitur stilometrik...")
df_raw_styled['uppercase_ratio'] = df_raw_styled['Email Text'].apply(
    lambda x: sum(1 for c in str(x) if c.isupper()) / (len(str(x)) + 1e-9) * 100
)
df_raw_styled['exclamation_count'] = df_raw_styled['Email Text'].apply(lambda x: str(x).count('!'))
df_raw_styled['dollar_count'] = df_raw_styled['Email Text'].apply(lambda x: str(x).count('$'))
df_raw_styled['digit_ratio'] = df_raw_styled['Email Text'].apply(
    lambda x: sum(1 for c in str(x) if c.isdigit()) / (len(str(x)) + 1e-9) * 100
)

# Visualisasi Boxplot 4 Fitur Stilometrik
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
axes = axes.flatten()

metrics = [
    ('uppercase_ratio', 'Uppercase Character Ratio (%)', 15),
    ('exclamation_count', 'Exclamation Mark Count (!)', 20),
    ('dollar_count', 'Dollar Currency Symbol Count ($)', 15),
    ('digit_ratio', 'Digit Character Ratio (%)', 25)
]

for idx, (col, title, cap_val) in enumerate(metrics):
    sub_plot = df_raw_styled[df_raw_styled[col] <= cap_val]
    sns.boxplot(data=sub_plot, x='Class_Name', y=col,
                palette={'Safe Email (0)': '#3498db', 'Phishing Email (1)': '#e74c3c'},
                ax=axes[idx], showmeans=True,
                meanprops={'marker':'o', 'markerfacecolor':'white', 'markeredgecolor':'black', 'markersize':'6'})
    axes[idx].set_title(title, fontsize=12, fontweight='bold')
    axes[idx].set_xlabel('')

plt.suptitle('Stylometric & Intimidation Syntax Analysis: Safe vs Phishing Emails', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
eda_file5 = EDA_PLOT_DIR / 'eda_05_stylometric_analysis.png'
plt.savefig(eda_file5, dpi=200, bbox_inches='tight')
plt.show()
print(f"Plot disimpan di: {eda_file5}")

# Tabel Rata-rata & Median Stilometrik
stylometric_summary = df_raw_styled.groupby('Class_Name')[['uppercase_ratio', 'exclamation_count', 'dollar_count', 'digit_ratio']].agg(['mean', 'median']).round(2).T
print("Perbandingan Statistik Stilometrik:")
display(stylometric_summary)
```

## 3.8 Analisis Trigram (Frasa 3 Kata Serangan Phishing)

Analisis n-gram tingkat 3 (*trigram*) memberikan pemahaman paling spesifik terhadap pola template phishing yang terdiri dari urutan 3 kata berturut-turut.


```python
def extract_top_trigrams(corpus, top_n=10):
    vec = CountVectorizer(ngram_range=(3, 3), stop_words=list(custom_stopwords), min_df=4, token_pattern=r'\b[a-z]{3,}\b')
    X_counts = vec.fit_transform(corpus)
    word_freq = X_counts.sum(axis=0)
    words = [(word, word_freq[0, idx]) for word, idx in vec.vocabulary_.items()]
    words = sorted(words, key=lambda x: x[1], reverse=True)
    return pd.DataFrame(words[:top_n], columns=['Trigram Phrase', 'Frequency'])

print("Mengekstraksi trigram...")
safe_trigrams = extract_top_trigrams(safe_corpus, top_n=10)
phish_trigrams = extract_top_trigrams(phish_corpus, top_n=10)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

axes[0].barh(safe_trigrams['Trigram Phrase'][::-1], safe_trigrams['Frequency'][::-1],
             color='#3498db', edgecolor='black', alpha=0.85)
axes[0].set_title('Top Trigrams in Safe Emails\n(Corporate Internal Operations)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Phrase Frequency')

axes[1].barh(phish_trigrams['Trigram Phrase'][::-1], phish_trigrams['Frequency'][::-1],
             color='#e74c3c', edgecolor='black', alpha=0.85)
axes[1].set_title('Top Trigrams in Phishing Emails\n(Targeted Lures & Action Prompts)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Phrase Frequency')

plt.suptitle('Trigram Phrase Comparison: Precise Social Engineering Sequences', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
eda_file6 = EDA_PLOT_DIR / 'eda_06_top_trigrams_comparison.png'
plt.savefig(eda_file6, dpi=200, bbox_inches='tight')
plt.show()
print(f"Plot disimpan di: {eda_file6}")
```

## 3.9 Proyeksi Semantik 2D (Latent Semantic Analysis via TruncatedSVD)

Untuk memahami bagaimana dokumen terdistribusi dalam ruang vektor TF-IDF, kita menggunakan Latent Semantic Analysis (LSA) berbasis `TruncatedSVD`:
1. Korpus diubah menjadi matriks TF-IDF ($5.000$ fitur unigram dan bigram).
2. Matriks dekomposisi direduksi ke 2 komponen semantik utama (*Latent Semantic Dimensions*).
3. Mengamati apakah dokumen Safe dan Phishing membentuk manifold semantik yang dapat dipisahkan oleh Linear SVM.


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

# Konstruksi TF-IDF pada sampel representatif untuk analisis semantik ruang laten
sample_text_df = df_clean_email.sample(10000, random_state=42)
tfidf_svd = TfidfVectorizer(max_features=5000, ngram_range=(1, 2), stop_words='english', sublinear_tf=True)
X_tfidf_sample = tfidf_svd.fit_transform(sample_text_df['cleaned_text'])

# Reduksi ke 2D menggunakan TruncatedSVD
svd_2d = TruncatedSVD(n_components=2, random_state=42)
X_svd_2d = svd_2d.fit_transform(X_tfidf_sample)

df_svd = pd.DataFrame(X_svd_2d, columns=['Semantic_Dim_1', 'Semantic_Dim_2'])
df_svd['Class_Name'] = sample_text_df['Class_Name'].values

plt.figure(figsize=(11, 6.5))
sns.scatterplot(data=df_svd, x='Semantic_Dim_1', y='Semantic_Dim_2', hue='Class_Name',
                palette={'Safe Email (0)': '#3498db', 'Phishing Email (1)': '#e74c3c'},
                alpha=0.5, s=22)
plt.title('Latent Semantic Analysis (TruncatedSVD 2D Projection on TF-IDF Space)', fontsize=13, fontweight='bold', pad=12)
plt.xlabel('Latent Semantic Dimension 1')
plt.ylabel('Latent Semantic Dimension 2')
plt.xlim(-0.02, 0.45)
plt.ylim(-0.25, 0.40)
plt.tight_layout()
eda_file7 = EDA_PLOT_DIR / 'eda_07_semantic_projection_svd.png'
plt.savefig(eda_file7, dpi=200, bbox_inches='tight')
plt.show()
print(f"Plot disimpan di: {eda_file7}")

print(f"Varians semantik dijelaskan oleh Komponen 1 dan 2: {sum(svd_2d.explained_variance_ratio_)*100:.2f}%")
print("Diagnosa Semantik: Dokumen Safe Email mengelompok erat pada poros semantik korporat, sedangkan Phishing Email tersebar luas di seluruh ruang laten karena keberagaman tipe penipuan (perbankan, hadiah, suspensi akun).")
```

# CHAPTER 4 : Text Vectorization & NLP Feature Pipeline

Pada bab ini, kita merancang arsitektur vektorisasi korpus teks menggunakan skema pembobotan statistik TF-IDF (*Term Frequency - Inverse Document Frequency*) dan membagi data secara terisolasi:

1. **Stratified Train-Test Split (80:20):** Membagi dataset bersih ke dalam 80% data latih (training set) dan 20% data uji independen (testing set) dengan stratifikasi target `is_phishing` untuk menjaga proporsi kelas yang identik.
2. **Pencegahan Kebocoran Informasi (No Data Leakage):** Vektorizer hanya di-*fit* pada subset training (`X_train`), kemudian digunakan untuk mentransformasikan subset testing (`X_test`) tanpa menyerap statistik dari data pengujian.
3. **Konfigurasi TF-IDF Vectorizer Berbasis Temuan EDA:**
   * `ngram_range=(1, 2)`: Menangkap ketergantungan leksikal unigram dan bigram (misalnya frasa pemicu seperti *verify account*, *click here*, *act now*).
   * `sublinear_tf=True`: Menerapkan penskalaan logaritmik $1 + \log(TF)$ untuk meredam dominasi frekuensi kata ekstrem.
   * `min_df=5`: Mengeliminasi kata langka (*rare tokens*) atau salah ketik leksikal yang muncul kurang dari 5 dokumen.
   * `max_df=0.85`: Mengeliminasi token korpus umum yang muncul di lebih dari 85% dokumen.
   * `max_features=25000`: Membatasi dimensi ruang fitur untuk menjaga efisiensi memori dan latensi inferensi.


```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib
import warnings
warnings.filterwarnings('ignore')

CLEAN_EMAIL_PATH = Path('../data/email/processed/email_cleaned.csv')
MODEL_DIR = Path('../models/email')
MODEL_DIR.mkdir(parents=True, exist_ok=True)
EVAL_PLOT_DIR = Path('../plots/email')
EVAL_PLOT_DIR.mkdir(parents=True, exist_ok=True)

df_clean = pd.read_csv(CLEAN_EMAIL_PATH)

# Pastikan tidak ada nilai null pada kolom teks
df_clean['cleaned_text'] = df_clean['cleaned_text'].fillna('')

X_raw = df_clean['cleaned_text']
y = df_clean['is_phishing']

print(f"Total Sampel Korpus Bersih: {len(df_clean):,} baris")
print(f"Distribusi Kelas: Safe (0) = {sum(y==0):,} ({sum(y==0)/len(y)*100:.2f}%), Phishing (1) = {sum(y==1):,} ({sum(y==1)/len(y)*100:.2f}%)")

# 1. Stratified Train-Test Split (80:20)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print(f"\nUkuran Subset Training: {len(X_train_raw):,} sampel")
print(f"Ukuran Subset Testing:  {len(X_test_raw):,} sampel")

# 2. Vektorisasi TF-IDF (Fit hanya pada X_train)
tfidf = TfidfVectorizer(
    ngram_range=(1, 2),
    sublinear_tf=True,
    min_df=5,
    max_df=0.85,
    max_features=25000
)

print("\nMelakukan fitting TF-IDF Vectorizer pada subset training...")
X_train = tfidf.fit_transform(X_train_raw)
X_test = tfidf.transform(X_test_raw)

print(f"Dimensi Matriks Fitur X_train: {X_train.shape[0]:,} baris x {X_train.shape[1]:,} fitur n-gram")
print(f"Dimensi Matriks Fitur X_test:  {X_test.shape[0]:,} baris x {X_test.shape[1]:,} fitur n-gram")
print(f"Sparsitas Matriks Fitur: {100 * (1 - X_train.nnz / (X_train.shape[0] * X_train.shape[1])):.2f}%")

# Simpan objek vectorizer untuk kebutuhan deployment inferensi
vectorizer_path = MODEL_DIR / 'tfidf_vectorizer_email.joblib'
joblib.dump(tfidf, vectorizer_path)
print(f"TF-IDF Vectorizer berhasil disimpan ke: {vectorizer_path}")
```

# CHAPTER 5 : Model Training & Baseline Comparison

Pada bab ini, kita membangun dan membandingkan dua model klasifikasi teks:
1. **Baseline Model:** `Logistic Regression` dengan regularisasi L2 (`C=1.0`), penyesuaian bobot kelas `class_weight='balanced'`, dan solver L-BFGS. Model ini menjadi tolok ukur standar dalam klasifikasi teks biner.
2. **Kandidat Model Produksi:** `Linear Support Vector Machine (LinearSVC)` yang dioptimalkan untuk ruang fitur berdimensi tinggi (*high-dimensional sparse representations*). Model ini kemudian dikalibrasi probabilitasnya menggunakan `CalibratedClassifierCV` (metode Platt Scaling sigmoid) agar dapat menghasilkan skor keyakinan probabilitas (*confidence scores*) untuk thresholding adaptif.
3. **Validasi Silang Terstratifikasi (Stratified 5-Fold Cross-Validation):** Mengukur stabilitas metrik generalisasi (Accuracy, Precision, Recall, F1-Score, ROC-AUC) dan efisiensi waktu pelatihan pada subset data latih.


```python
import time
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 1. Inisialisasi Model Baseline (Logistic Regression)
lr_baseline = LogisticRegression(
    C=1.0,
    class_weight='balanced',
    max_iter=1000,
    random_state=42,
    solver='lbfgs',
    n_jobs=2
)

# 2. Inisialisasi Model Produksi (Linear SVM dengan Kalibrasi Probabilitas)
base_svc = LinearSVC(
    C=1.0,
    class_weight='balanced',
    random_state=42,
    max_iter=2000
)
svm_production = CalibratedClassifierCV(
    estimator=base_svc,
    method='sigmoid',
    cv=3
)

# 3. Pelatihan Model pada Full Training Set
print("1. Melatih Baseline Logistic Regression...")
t0 = time.time()
lr_baseline.fit(X_train, y_train)
lr_fit_time = time.time() - t0
print(f"Logistic Regression selesai dilatih dalam {lr_fit_time:.2f} detik.")

print("\n2. Melatih Production Calibrated Linear SVM...")
t0 = time.time()
svm_production.fit(X_train, y_train)
svm_fit_time = time.time() - t0
print(f"Calibrated Linear SVM selesai dilatih dalam {svm_fit_time:.2f} detik.")

# 4. Simpan Model Terlatih ke Disk
lr_model_path = MODEL_DIR / 'lr_baseline_email.joblib'
svm_model_path = MODEL_DIR / 'svm_production_email.joblib'

joblib.dump(lr_baseline, lr_model_path)
joblib.dump(svm_production, svm_model_path)
print(f"\nModel berhasil disimpan ke:")
print(f"- {lr_model_path}")
print(f"- {svm_model_path}")
```

### Perbandingan Validasi Silang (Stratified 5-Fold Cross Validation)

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = {'accuracy': 'accuracy', 'precision': 'precision', 'recall': 'recall', 'f1': 'f1', 'roc_auc': 'roc_auc'}

print("Menjalankan Stratified 5-Fold Cross Validation pada subset training...")
lr_cv_model = LogisticRegression(C=1.0, class_weight='balanced', max_iter=1000, random_state=42, n_jobs=1)
svm_cv_model = CalibratedClassifierCV(estimator=LinearSVC(C=1.0, class_weight='balanced', random_state=42, max_iter=2000), method='sigmoid', cv=3)

cv_lr = cross_validate(lr_cv_model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=2)
cv_svm = cross_validate(svm_cv_model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=2)

cv_comparison = pd.DataFrame({
    'Metric (5-Fold CV Mean)': ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC', 'Fit Time (s)'],
    'Logistic Regression (Baseline)': [
        f"{cv_lr['test_accuracy'].mean():.4f} +/- {cv_lr['test_accuracy'].std():.4f}",
        f"{cv_lr['test_precision'].mean():.4f} +/- {cv_lr['test_precision'].std():.4f}",
        f"{cv_lr['test_recall'].mean():.4f} +/- {cv_lr['test_recall'].std():.4f}",
        f"{cv_lr['test_f1'].mean():.4f} +/- {cv_lr['test_f1'].std():.4f}",
        f"{cv_lr['test_roc_auc'].mean():.4f} +/- {cv_lr['test_roc_auc'].std():.4f}",
        f"{cv_lr['fit_time'].mean():.2f}s"
    ],
    'Linear SVM (Production)': [
        f"{cv_svm['test_accuracy'].mean():.4f} +/- {cv_svm['test_accuracy'].std():.4f}",
        f"{cv_svm['test_precision'].mean():.4f} +/- {cv_svm['test_precision'].std():.4f}",
        f"{cv_svm['test_recall'].mean():.4f} +/- {cv_svm['test_recall'].std():.4f}",
        f"{cv_svm['test_f1'].mean():.4f} +/- {cv_svm['test_f1'].std():.4f}",
        f"{cv_svm['test_roc_auc'].mean():.4f} +/- {cv_svm['test_roc_auc'].std():.4f}",
        f"{cv_svm['fit_time'].mean():.2f}s"
    ]
})

print("Tabel Komparasi Performa Cross-Validation (Training Set):")
display(cv_comparison)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, precision_recall_curve, auc

# 1. Prediksi pada Independent Testing Set
y_pred_lr = lr_baseline.predict(X_test)
y_prob_lr = lr_baseline.predict_proba(X_test)[:, 1]

y_pred_svm = svm_production.predict(X_test)
y_prob_svm = svm_production.predict_proba(X_test)[:, 1]

print("=== CLASSIFICATION REPORT: LOGISTIC REGRESSION (BASELINE) ===")
print(classification_report(y_test, y_pred_lr, target_names=['Safe Email (0)', 'Phishing Email (1)'], digits=4))

print("=== CLASSIFICATION REPORT: LINEAR SVM (PRODUCTION) ===")
print(classification_report(y_test, y_pred_svm, target_names=['Safe Email (0)', 'Phishing Email (1)'], digits=4))
```

## 6.3 Analisis Kesalahan Prediksi (Error Analysis) & Kasus Gagal

Inspeksi sampel False Positive (email sah terblokir) dan False Negative (email phishing lolos ke kotak masuk) pada model Linear SVM untuk memitigasi risiko operasional.


```python
test_indices = X_test_raw.index
df_error = df_clean.loc[test_indices].copy()
df_error['pred_lr'] = y_pred_lr
df_error['pred_svm'] = y_pred_svm
df_error['prob_svm'] = y_prob_svm

# 1. False Positives: Safe Email (0) diprediksi sebagai Phishing (1)
fp_svm = df_error[(df_error['is_phishing'] == 0) & (df_error['pred_svm'] == 1)]
print(f"Jumlah False Positive (Linear SVM): {len(fp_svm)} sampel dari {sum(y_test==0):,} email sah ({len(fp_svm)/sum(y_test==0)*100:.3f}%)")
if len(fp_svm) > 0:
    print("\nContoh Sampel False Positive (Email Sah yang Terindikasi Phishing):")
    for idx, row in fp_svm.head(3).iterrows():
        snippet = row['cleaned_text'][:160].replace('\n', ' ')
        print(f"- [ID {idx}] Prob: {row['prob_svm']:.4f} | Snippet: {snippet}...")

# 2. False Negatives: Phishing Email (1) diprediksi sebagai Safe (0)
fn_svm = df_error[(df_error['is_phishing'] == 1) & (df_error['pred_svm'] == 0)]
print(f"\nJumlah False Negative (Linear SVM): {len(fn_svm)} sampel dari {sum(y_test==1):,} email phishing ({len(fn_svm)/sum(y_test==1)*100:.3f}%)")
if len(fn_svm) > 0:
    print("\nContoh Sampel False Negative (Email Phishing yang Lolos):")
    for idx, row in fn_svm.head(3).iterrows():
        snippet = row['cleaned_text'][:160].replace('\n', ' ')
        print(f"- [ID {idx}] Prob: {row['prob_svm']:.4f} | Snippet: {snippet}...")
```

## 6.4 Kesimpulan Operasional & Rekomendasi Deployment

1. **Pemilihan Model Produksi:** `Linear SVM (dikalibrasi dengan sigmoid/Platt scaling)` mengungguli Logistic Regression dalam akurasi dan pemisahan hyperplane berdimensi tinggi, serta membuktikan efisiensi komputasi yang sangat cepat (inferensi sub-milidetik per email).
2. **Threshold Kalibrasi:** 
   * Default threshold `0.50` memberikan trade-off seimbang antara precision dan recall.
   * Untuk deployment pada Email Security Gateway (SEG) perusahaan, threshold pemblokiran otomatis dapat diatur pada probabilitas `> 0.80`, sementara email dengan skor `0.45 - 0.80` diberi label peringatan peringatan keamanan (*security warning banner*) atau dikarantina untuk verifikasi manual oleh analis SOC.
3. **Pertahanan Berlapis (*Defense-in-Depth*):** Model klasifikasi teks TF-IDF + SVM ini sangat efektif dipadukan dengan modul URL Phishing Detection (`url.ipynb` dengan XGBoost). Ketika sebuah email mengandung tautan mencurigakan, skor ancaman gabungan dihitung secara ansambel untuk memberikan perlindungan komprehensif.
